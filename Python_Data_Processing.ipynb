```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

deliveries = pd.read_csv("deliveries.csv")

deliveries.head()
deliveries.info()

deliveries.isnull().sum()

deliveries.fillna(0, inplace=True)
deliveries.isnull().sum()

deliveries.drop_duplicates(inplace=True)

deliveries['delivery_status'].value_counts()

deliveries['delivery_status'].value_counts().plot(kind='bar')

plt.title("Delivery Status Analysis")
plt.xlabel("Delivery Status")
plt.ylabel("Number of Deliveries")

plt.show()

deliveries['customer_rating_post_delivery'].describe()


plt.hist(deliveries['customer_rating_post_delivery'], bins=10)

plt.title("Customer Rating Distribution")
plt.xlabel("Customer Rating")
plt.ylabel("Frequency")

plt.show()

plt.scatter(deliveries['route_distance_km'],
            deliveries['fuel_or_charge_cost'])

plt.title("Route Distance vs Fuel Cost")
plt.xlabel("Route Distance (KM)")
plt.ylabel("Fuel or Charging Cost")

plt.show()

deliveries['manual_route_override_count'].value_counts()

deliveries['manual_route_override_count'].value_counts().plot(kind='bar')

plt.title("Manual Route Override Analysis")
plt.xlabel("Route Override Count")
plt.ylabel("Frequency")

plt.show()

deliveries['fuel_or_charge_cost'].describe()

plt.boxplot(deliveries['fuel_or_charge_cost'])

plt.title("Fuel or Charging Cost Distribution")
plt.ylabel("Fuel or Charging Cost")

plt.show()

deliveries.corr(numeric_only=True)
```
